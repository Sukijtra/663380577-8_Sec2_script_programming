# วันที่ 1 : ออกแบบและพัฒนาระบบ

## งานที่ทำ

* ออกแบบโครงสร้างระบบ POS
* กำหนดข้อมูลเมนูเครื่องดื่ม
* พัฒนาฟังก์ชันแสดงเมนู
* พัฒนาฟังก์ชันรับออเดอร์
* เพิ่มการเลือกขนาดแก้ว ประเภทนม และระดับความหวาน

## การใช้ AI

**จำนวน Prompt ที่ใช้:** ประมาณ 15 Prompt

ตัวอย่าง Prompt

* ช่วยออกแบบระบบ POS ร้านกาแฟด้วย Python
* วิธีใช้ Dictionary เก็บข้อมูลเมนู
* วิธีแบ่งโปรแกรมเป็นฟังก์ชัน
* วิธีตรวจสอบข้อมูลที่ผู้ใช้กรอกด้วย try-except
* วิธีคำนวณราคาจากตัวเลือกหลายรายการ

## สิ่งที่ได้เรียนรู้

* การออกแบบโครงสร้างโปรแกรมด้วยฟังก์ชัน
* การใช้ Dictionary และ List เพื่อจัดเก็บข้อมูล
* การตรวจสอบข้อมูลที่ผู้ใช้ป้อน
* การคำนวณราคาจากหลายเงื่อนไข

## ปัญหาที่พบ

* การคำนวณราคาจากขนาดแก้วและประเภทนมมีความซับซ้อน
* ผู้ใช้กรอกข้อมูลไม่ถูกต้องทำให้โปรแกรมเกิดข้อผิดพลาด

## วิธีแก้ไข

* แยกการคำนวณราคาออกเป็นขั้นตอน
* เพิ่มการตรวจสอบข้อมูลด้วย while และ try-except

# วันที่ 2 : พัฒนาระบบและทดสอบ

## งานที่ทำ

* พัฒนาระบบสมาชิก
* เพิ่มระบบส่วนลด
* พัฒนาระบบชำระเงิน
* เพิ่มระบบคำนวณเงินทอน
* เพิ่มการออกใบเสร็จ
* เพิ่มระบบสรุปยอดขายประจำวัน
* ทดสอบและแก้ไขข้อผิดพลาดของโปรแกรม

## การใช้ AI

**จำนวน Prompt ที่ใช้:** ประมาณ 18 Prompt

ตัวอย่าง Prompt

* วิธีสร้างใบเสร็จด้วย Python
* วิธีคำนวณเงินทอน
* วิธีบันทึกประวัติการขายใน List
* วิธีสรุปยอดขายประจำวัน
* ช่วยตรวจสอบและอธิบายโค้ดแต่ละฟังก์ชัน

## สิ่งที่ได้เรียนรู้

* การสร้างระบบชำระเงิน
* การจัดเก็บข้อมูลธุรกรรม
* การสรุปข้อมูลจาก List
* การปรับปรุงโค้ดให้เป็นระเบียบและอ่านง่าย

## ปัญหาที่พบ

* การแสดงผลใบเสร็จไม่เป็นระเบียบ
* การคำนวณยอดขายรวมต้องวนลูปหลายชั้น

## วิธีแก้ไข

* ปรับรูปแบบการแสดงผลด้วย f-string
* ใช้ฟังก์ชัน sum() และการวนลูปเพื่อคำนวณยอดขาย
