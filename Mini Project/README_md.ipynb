# ระบบ POS ร้านกาแฟ (Coffee Shop POS System)

## รายละเอียดโครงงาน

โครงงานนี้เป็นระบบ POS (Point of Sale) สำหรับร้านกาแฟ พัฒนาด้วยภาษา Python บน Google Colab เพื่อช่วยในการจัดการการขายหน้าร้าน ตั้งแต่การรับออเดอร์ คำนวณราคา รับชำระเงิน ออกใบเสร็จ และสรุปยอดขายประจำวัน

## ความสามารถของระบบ

- แสดงเมนูเครื่องดื่ม
- เลือกขนาดแก้ว (S,M,L)
- เลือกประเภทนม
- เลือกระดับความหวาน
- สั่งซื้อได้หลายรายการ
- ระบบสมาชิกและส่วนลด 10%
- คำนวณราคารวมอัตโนมัติ
- รับชำระเงินสด
- คำนวณเงินทอน
- พิมพ์ใบเสร็จ
- สรุปยอดขายประจำวัน

---

## วิธีใช้งาน

1. เปิดไฟล์ `POS_Cafe.ipynb` ด้วย Google Colab
2. กด Run All
3. เลือกเมนูที่ต้องการจากหน้าหลัก
4. ทำรายการสั่งซื้อ
5. รับชำระเงินและออกใบเสร็จ
6. ดูสรุปยอดขายประจำวันได้จากเมนูหลัก
