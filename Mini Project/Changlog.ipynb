# บันทึกการเปลี่ยนแปลง (CHANGELOG)

### เพิ่ม (Added)
- สร้างโครงสร้างโปรแกรมระบบ POS ร้านกาแฟ
- เพิ่มข้อมูลเมนูเครื่องดื่ม

- พัฒนาฟังก์ชันเลือกขนาดแก้ว ประเภทนม และระดับความหวาน get_customization
