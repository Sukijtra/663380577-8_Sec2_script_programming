Part 1: Create a Custom Module

```python
%%writefile my_utils.py

def greet(name):
    """แสดงข้อความทักทาย"""
    print(f"Hello, {name}!")


def is_prime(number):
    """ตรวจสอบว่าตัวเลขเป็นจำนวนเฉพาะหรือไม่"""
    if number < 2:
        return False

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True
```

เป็นการสร้าง Custom Module ของตัวเอง โดยสร้างไฟล์ชื่อ my_utils.py ด้วยคำสั่ง %%writefile

ภายใน Module ประกอบด้วย 2 ฟังก์ชัน ได้แก่

greet(name) ใช้สำหรับแสดงข้อความทักทาย โดยรับชื่อผ่าน Parameter name และใช้ print() เพื่อแสดงข้อความ
is_prime(number) ใช้สำหรับตรวจสอบว่าตัวเลขที่รับเข้ามาเป็นจำนวนเฉพาะหรือไม่ โดยจะคืนค่า True ถ้าเป็นจำนวนเฉพาะ และคืนค่า False ถ้าไม่ใช่

ฟังก์ชัน is_prime() จะตรวจสอบตัวเลขตั้งแต่ 2 ไปจนถึงค่ารากที่สองของตัวเลขนั้น หากหารลงตัว แสดงว่าไม่ใช่จำนวนเฉพาะ

การสร้างฟังก์ชันไว้ในไฟล์ my_utils.py ทำให้เราสามารถนำ Module นี้ไป Import และใช้งานในโปรแกรมอื่นได้ ซึ่งเป็นแนวคิดสำคัญของการแบ่งโค้ดออกเป็นส่วน ๆ และนำกลับมาใช้ซ้ำ

Part 2: Use the Custom Module and Standard Modules

```python
import my_utils
import math
import random

# เรียกใช้ฟังก์ชัน greet()
my_utils.greet("Alice")

# ทดสอบฟังก์ชัน is_prime()
print("7 is prime:", my_utils.is_prime(7))
print("10 is prime:", my_utils.is_prime(10))

# ใช้ math.sqrt() หารากที่สอง
number = 25
print("Square root of", number, "=", math.sqrt(number))

# ใช้ random.randint() สุ่มตัวเลข 1-100
random_number = random.randint(1, 100)
print("Random number:", random_number)
```

เป็นการนำ Custom Module ที่สร้างจาก Part 1 มาใช้งานร่วมกับ Standard Library Modules ของ Python

เริ่มจากใช้ import my_utils เพื่อ Import Module ที่เราสร้างขึ้นเอง จากนั้นสามารถเรียกใช้ฟังก์ชันภายใน Module ด้วยรูปแบบ my_utils.function_name()

ตัวอย่างเช่น my_utils.greet("Alice") ใช้เรียกฟังก์ชัน greet() เพื่อแสดงข้อความทักทาย และ my_utils.is_prime(7) กับ my_utils.is_prime(10) ใช้ตรวจสอบว่าตัวเลขเป็นจำนวนเฉพาะหรือไม่

ต่อมา Import Module มาตรฐานของ Python ได้แก่ math และ random

ใช้ math.sqrt(25) เพื่อคำนวณรากที่สองของ 25 ซึ่งได้ผลลัพธ์เป็น 5.0

และใช้ random.randint(1, 100) เพื่อสุ่มจำนวนเต็มตั้งแต่ 1 ถึง 100 โดยค่าที่สุ่มได้อาจแตกต่างกันในแต่ละครั้งที่รันโปรแกรม
