#**Lab 6.1: Simple Calculator Functions**

```python
# ============================================================
# Lab 6.1: Simple Calculator Functions
# ============================================================

# Function: Addition
def add(a, b):
    """คำนวณผลบวกของตัวเลขสองจำนวน"""
    return a + b


# Function: Subtraction
def subtract(a, b):
    """คำนวณผลลบของตัวเลขสองจำนวน"""
    return a - b


# Function: Multiplication
def multiply(a, b):
    """คำนวณผลคูณของตัวเลขสองจำนวน"""
    return a * b


# Function: Division
def divide(a, b):
    """คำนวณผลหารของตัวเลขสองจำนวน"""
    if b == 0:
        return "Error: Division by zero"
    return a / b


# Challenge: Power
def power(base, exponent=2):
    """คำนวณเลขยกกำลัง โดยค่าเริ่มต้นของ exponent คือ 2"""
    return base ** exponent


# ============================================================
# Main Program
# ============================================================

while True:

    print("\n===== Simple Calculator =====")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Power")
    print("6. Exit")
    print("=============================")

    choice = input("Choose an operation (1-6): ")

    # Exit
    if choice == "6":
        print("Thank you for using Simple Calculator!")
        break

    # Check valid menu
    if choice not in ["1", "2", "3", "4", "5"]:
        print("Invalid choice. Please try again.")
        continue

    # Get numbers
    try:
        num1 = float(input("Enter first number: "))

        if choice == "5":
            exponent_input = input(
                "Enter exponent (press Enter for 2): "
            )

            if exponent_input == "":
                result = power(num1)
            else:
                exponent = float(exponent_input)
                result = power(num1, exponent)

        else:
            num2 = float(input("Enter second number: "))

            # Call appropriate function
            if choice == "1":
                result = add(num1, num2)

            elif choice == "2":
                result = subtract(num1, num2)

            elif choice == "3":
                result = multiply(num1, num2)

            elif choice == "4":
                result = divide(num1, num2)

        # Display result
        print("Result:", result)

    except ValueError:
        print("Error: Please enter numbers only.")
```

เป็นการใช้งาน Function, Parameters, Arguments, Return Value, Default Parameters, while loop, if-elif และ try-except รวมถึงการแบ่งโปรแกรมออกเป็นฟังก์ชันย่อย ๆ ซึ่งช่วยให้โค้ดเป็นระเบียบ อ่านง่าย แก้ไขง่าย และสามารถนำกลับมาใช้ซ้ำได้
