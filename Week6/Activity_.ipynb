# **Activity 6.1: Refactoring a Previous Lab Assignment**

ตัวอย่างนี้เลือก Week 2: Number Classifier

```python
# Activity 6.1: Refactoring Number Classifier


def is_positive_negative_zero(num):
    """Determine whether a number is positive, negative, or zero."""

    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"


def is_even_odd(num):
    """Determine whether a number is even or odd."""

    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"


# Main program

print("===== Number Classifier =====")

number = int(input("Enter a number: "))

result_sign = is_positive_negative_zero(number)
result_parity = is_even_odd(number)

print("Number:", number)
print("Classification:", result_sign)
print("Parity:", result_parity)
```

อธิบายการ Refactoring

เดิมการตรวจสอบตัวเลขอาจเขียนทุกอย่างรวมกันในส่วนหลักของโปรแกรม แต่หลังจาก Refactoring ได้แบ่งออกเป็น 2 Functions ได้แก่

```python
is_positive_negative_zero(num)
```

ใช้ตรวจสอบว่าเป็น Positive, Negative หรือ Zero และ

```python
is_even_odd(num)
```

ใช้ตรวจสอบว่าเป็น Even หรือ Odd

ทำให้โค้ดมีความเป็น Modular มากขึ้น และสามารถนำ Functions กลับมาใช้ซ้ำได้

# **Activity 6.2: Discussion: Benefits of Modular Programming &amp; Standard Library Exploration**

1.  Benefits of Modularity

คำตอบ:Large Projects
การแบ่งโปรแกรมขนาดใหญ่ออกเป็น Functions และ Modules ช่วยให้โค้ดเป็นส่วน ๆ และจัดการได้ง่ายขึ้น
